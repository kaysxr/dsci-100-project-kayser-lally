#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
library(tidyverse)

tidied_bball_data <- womens_bball_data |>
    
    # select the columns needed: year, school, chosen predictors (seed, reg_wins_pct, conf_wins_pct), and the target variable (tourney_finish) 
    select(year, school, seed, reg_wins_pct, conf_wins_pct, tourney_finish) |>
    
    # remove rows with missing values in the key variables
    drop_na(seed, reg_wins_pct, conf_wins_pct, tourney_finish) |>
   
    # transform target variable into an ordered factor to create a meaningful sequence and forced hierarchy
    mutate(tourney_finish = factor(tourney_finish, levels = c("first_round_loss", "second_round_loss", "top_16_loss", "top_8_loss", "top_4_loss", "top_2_loss", "champ")))

head(tidied_bball_data, 6)
```

**VISUALIZATION 1**

This box plot directly evaluates the relationship between a team's tournament seed and its final tourney finish. By mapping seed on the vertical axis and finish categories on the horizontal axis, we can easily assess whether lower seed numbers (higher ranked teams) consistently correspond to deeper tournament runs (like reaching the championship, or top 4). This visualization provides insightt into whether seed serves as a strong predictor for the classification model.

To predict tournament finish I will use a k-NN classification model with seed, reg_wins_pct, and conf_wins_pct as predictors. 

This method is appropriate because my goal is to predict a category (how far a team finishes in the tournament) by looking at past teams that are most similar in their stats. While exploratory analysis showed strong trends for seed and regular-season win percentage and a slightly more subtle pattern for conference win percentage, including all three lets the model evaluate them together.

A major limitation of k-NN is that it is sensitive to different number scales, meaning that variables with larger numbers can unfairly dominate the results if I'm not careful. It can also overfit the data if the choice of k isn’t right.  

To process the data, I will remove missing values and scale the numeric predictors so they have equal footing. Next, I will split the data into an 80/20 train-test split, using 80% of the data to train the model, and saving 20% to test how well it predicts outcomes for unfamiliar teams. 

Since my target variable is categorical, I can’t calculate a numerical error distance like RMSE. To assess performance, I will use classification accuracy and misclassification error rate upon my test set. The overall classification accuracy will tell the percentage of correct guesses my model makes, and the misclassification error rate will tell the percentage of incorrect guesses the model makes.
